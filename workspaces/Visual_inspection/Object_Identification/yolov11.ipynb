```python
import os
from collections import Counter

label_root = r"E:\sem7\FYP\9_24\objectdetetction\datasets\custom_merged_obb\labels\train"

line_counts = Counter()
for f in os.listdir(label_root):
    if not f.endswith(".txt"): continue
    with open(os.path.join(label_root, f)) as file:
        for line in file:
            n = len(line.strip().split())
            line_counts[n] += 1

print("Label line formats found:", dict(line_counts))
```

```python
import os

label_dir = r"E:\sem7\FYP\9_24\objectdetetction\datasets\custom_merged_obb\labels\train"

for fname in os.listdir(label_dir):
    if fname.endswith('.txt'):
        path = os.path.join(label_dir, fname)
        with open(path, 'r') as f:
            for i, line in enumerate(f):
                parts = line.strip().split()
                if len(parts) != 9:
                    print(f"{fname} line {i+1}: Expected 9 values, got {len(parts)}")
                else:
                    floats = list(map(float, parts[1:]))
                    if not all(0.0 <= v <= 1.0 for v in floats):
                        print(f"{fname} line {i+1}: Non-normalized values found: {floats}")
```
